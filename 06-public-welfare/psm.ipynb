```python
import hces
import pandas as pd
df = hces.load()
df.head()
```

## PMGKY

```python
df['treat'] = (df['has_benefited_from_pmgky'] == 'yes').astype(int)
```

```python
xdf = hces.propensity_score_match(df, num_cols=['family_size', 'n_children', 'n_elderly'],                                  
                                  cat_cols=['sector', 'nss_region', 'employed_annual', 'max_income_from', 'hoh_religion', 'caste'])
```

```python
ydf = hces.compare(xdf[xdf['support']], ['label', 'treat'], hces.FOOD_COLS)
vals = ydf.xs(1, level='treat') - ydf.xs(0, level='treat')
pc = vals / ydf.xs(0, level='treat')
pd.concat([vals.round(), pc.round(2)], axis=0, keys=['values', 'percentage'])
```

```python
xdf['treat'] = xdf['treat'].astype(int)
ydf = hces.compare(xdf[xdf['support']], ['label', 'treat'], hces.EDU_COLS)
vals = ydf.xs(1, level='treat') - ydf.xs(0, level='treat')
pc = vals / ydf.xs(0, level='treat')
pd.concat([vals.round(), pc.round(2)], axis=0, keys=['values', 'percentage'])
```

```python
xdf['treat'] = xdf['treat'].astype(int)
ydf = hces.compare(xdf[xdf['support']], ['label', 'treat'], hces.HEALTH_COLS)
vals = ydf.xs(1, level='treat') - ydf.xs(0, level='treat')
pc = vals / ydf.xs(0, level='treat')
pd.concat([vals.round(), pc.round(2)], axis=0, keys=['values', 'percentage'])
```
